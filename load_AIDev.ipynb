### Check PR type

```python
pr_task_type_df['type'].value_counts()
```

```python
# Identify PRs with modifications in specified build files
build_extensions = ['.gradle', '.gradle.kts', '.xml', '.cmake', 'Makefile','CMakeLists.txt']

def is_build_file(filename):

  if filename is None: # This line was added to handle the error
        return False
  for ext in build_extensions:
        if filename.endswith(ext):
            return True
  return False

build_file_mask = pr_commit_details_df['filename'].apply(is_build_file)
pr_with_build_changes_ids = pr_commit_details_df[build_file_mask]['pr_id'].unique()

# Filter the main PRs dataframe
pr_with_build_changes_df = pr_df[pr_df['id'].isin(pr_with_build_changes_ids)]

# Save the filtered data to a CSV file
pr_with_build_changes_df.to_csv('pr_with_build_changes.csv', index=False)

print("CSV file 'pr_with_build_changes.csv' created successfully.")
```

```python
display(pr_with_build_changes_df)
```

```python
# Merge pr_with_build_changes_df with pr_task_type_df to get the PR type
pr_with_details = pd.merge(pr_with_build_changes_df, pr_task_type_df[['id', 'type']], on='id', how='left')

# Aggregate all filenames for each PR from pr_commit_details_df
all_files_changed = pr_commit_details_df.groupby('pr_id')['filename'].agg(lambda x: list(x.dropna().unique())).reset_index()
all_files_changed.rename(columns={'pr_id': 'id', 'filename': 'all_changed_files'}, inplace=True)

# Merge the aggregated filenames with the pr_with_details dataframe
pr_with_details = pd.merge(pr_with_details, all_files_changed, on='id', how='left')

# Display the updated dataframe with 'type' and 'all_changed_files'
print("DataFrame with PR type and all changed files:")
display(pr_with_details)
pr_with_details.to_csv('pr_with_build_changes.csv', index=False)
```

```python
# Filter pr_commit_details_df to get only commits that changed build files
build_file_commits = pr_commit_details_df[build_file_mask]

# Merge the filtered commit details with the pr_with_details dataframe
commits_with_pr_details = pd.merge(build_file_commits, pr_with_details[['id', 'agent', 'type', 'all_changed_files']], left_on='pr_id', right_on='id', how='left')

# Drop the redundant 'id' column from the merge
commits_with_pr_details = commits_with_pr_details.drop(columns=['id'])

# Display the resulting dataframe
print("Commits with build file changes and associated PR details:")
display(commits_with_pr_details)
commits_with_pr_details.to_csv('commits_with_build_changes.csv', index=False)
```

```python
# Merge pr_with_build_changes_df with pr_task_type_df to get the PR type
pr_with_types = pd.merge(pr_with_build_changes_df, pr_task_type_df[['id', 'type']], on='id', how='left')

# Get the value counts for the 'type' column
type_counts = pr_with_types['type'].value_counts()

# Display the results
print("Occurrence of each PR type in PRs with build file changes:")
display(type_counts)
```
